```python
import pandas as pd

ds = pd.read_csv('Housing.csv')

ds.head()
```

```python
ds.columns = ds.columns.str.strip()
ds.columns
```

```python
ds["total_rooms"] = ds["bedrooms"] + ds["bathrooms"]
```

```python
amenities = [
    "mainroad",
    "guestroom",
    "basement",
    "hotwaterheating",
    "airconditioning",
    "prefarea"
]

for col in amenities:
    ds[col] = ds[col].map({"yes": 1, "no": 0})

ds["total_amenities"] = ds[amenities].sum(axis=1)
```

```python
ds["parking_per_bedroom"] = ds["parking"] / ds["bedrooms"]
```

```python
ds["luxury_score"] = (
    ds["airconditioning"] +
    ds["guestroom"] +
    ds["basement"] +
    ds["hotwaterheating"] +
    ds["prefarea"]
)
```

```python
ds["furnishingstatus"] = ds["furnishingstatus"].map({
    "unfurnished": 0,
    "semi-furnished": 1,
    "furnished": 2
})
```

```python
binary_columns = [
    "mainroad",
    "guestroom",
    "basement",
    "hotwaterheating",
    "airconditioning",
    "prefarea"
]

for col in binary_columns:
    ds[col] = ds[col].map({"yes": 1, "no": 0})


ds["furnishingstatus"] = ds["furnishingstatus"].map({
    "unfurnished": 0,
    "semi-furnished": 1,
    "furnished": 2
})


ds["total_rooms"] = ds["bedrooms"] + ds["bathrooms"]

ds["area_per_story"] = ds["area"] / ds["stories"]

ds["total_amenities"] = ds[binary_columns].sum(axis=1)

ds["parking_per_bedroom"] = ds["parking"] / ds["bedrooms"]

ds["luxury_score"] = (
    ds["airconditioning"] +
    ds["guestroom"] +
    ds["basement"] +
    ds["hotwaterheating"] +
    ds["prefarea"]
)

ds.head()
```
